## Ingestión de datos 

1. Importando Librerias

```python
# Importando librerias 
import pandas as pd
import openpyxl as ex
```

2. Importando datos

```python
df_xlsx = pd.read_excel('flota_europa.xlsx', sheet_name='registros')
df_csv = pd.read_csv('flota_norteamerica.csv')
df_text = pd.read_csv('flota_asia.txt', delimiter="~")
df_parquet = pd.read_parquet('flota_datalake.parquet', engine='pyarrow')
```

3. Concatenando datos

```python
df = pd.concat([df_xlsx, df_csv, df_text,df_parquet], ignore_index = True)
```

Preparación de Datos

```python
# Imprimiendo información del data frame
df.info()
```

```python
# Imprimiendo duplicados
df.duplicated().sum()
```

```python
# Imprimiendo valores nulos
df.isnull().sum()
```

```python
df.describe()
```

```python
#Calcunado la median de la columna
hp_median = df['horsepower'].median()

df['horsepower'] = df['horsepower'].fillna(hp_median)
```

```python
df.isna().sum()
```

Elminando Registros Duplicados

```python
print(f'Tamño de nuestro dataset antes de eleminar duplicados: {df.shape[0]}')

df = df.drop_duplicates().reset_index(drop = True)

print(f'Tamño de nuestro dataset antes de eleminar duplicados: {df.shape[0]}')
```
